```python
# Import libraries
import pandas as pd
import numpy as np
```

```python
# Import Raw of Dataset
df = pd.read_csv("India Agriculture Crop Production.csv")
df.head(5)
```

```python
# Check the info of each column (total row, column, null values)
df.info()
```

```python
# Check null values in the data
df.isna().sum()
```

```python
print("Total row BEFORE droping null values:", df.shape[0])

# Note
"""
If the Area is nothing (null), it won't producting any crop
So the rows with null Area should be removed
"""

df = df.dropna(subset=['Area'])
print("Total row AFTER droping null values:", df.shape[0])
```

```python
# Recheck null values after drop 1 column
df.isna().sum()
```

```python
# Note
null_prod = df['Production'].isna().sum()
print(f"""
      There are no null values in all column, except 'Production'.
      There are {null_prod} null values, that means no crop is produced.
      You can replace it with 0 value, so that the data is completely clean from null values.
      But you can ignore it (this will have no effect).
      """)

# Replace null values of 'Production' to be 0 value
df['Production'] = df['Production'].fillna(0)
```

```python
# Recheck null values (final)
df.isna().sum()
```

```python
# Check duplicate values
duplicate = df.duplicated().sum()
if duplicate == 0:
    print("The duplicate values is 0, nothing to remove")
else:
    print(f"The duplicate values is {duplicate}, go remove it")
```

```python
# Check all of the Year is complete
df['Year'].unique()
```

```python
# Remove 3 back characters of 'Year'
df['Year'] = df['Year'].str.slice(0, -3)
```

```python
# Changing all values of 'District' to Title Case
df['District'] = df['District'].str.title()
```

```python
# Round down all values of 'Yield' to 2 digits after decimal
df['Yield'] = df['Yield'].round(2)
```

```python
# Convert all unit of the 'Production' values to Tonne
"""
1 Tonne = 1 Tonne
1 Nut = 0.0001 Tonne
1 Bale = 0.22 Tonne
"""

df['Production in Tonnes'] = df.apply(lambda row:
                                        row['Production'] if row['Production Units'] == 'Tonnes'
                                        else row['Production'] * 0.0001 if row['Production Units'] == 'Nuts'
                                        else row['Production'] * 0.22 if row['Production Units'] == 'Bales'
                                        else None, axis=1)
```

```python
# Assign the clean data into a new variable
df_clean = df
df_clean
```

```python
# Save the clean data to Excel
df.to_excel("clean_Dataset.xlsx", index = False)
```
